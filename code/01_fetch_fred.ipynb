# 01 Fetch data from FRED

Download the nine FRED series to `data/raw/`, one CSV each at their own native frequency. 


```python
from pathlib import Path
import pandas as pd
from dotenv import load_dotenv
import os
from fredapi import Fred
ROOT = Path.cwd().parent
```

```python
load_dotenv(ROOT / ".env")
key = os.getenv("FRED_API_KEY")
fred = Fred(api_key=key)
```

```python
series_ids = [
    "MORTGAGE30US",         # 30-year fixed mortgage rate, weekly
    "DGS10",                # 10-year Treasury constant maturity, daily
    "DFII10",               # 10-year TIPS yield, daily
    "T10YIE",               # 10-year breakeven inflation, daily
    "WSHOMCB",              # Fed holdings of agency MBS, weekly
    "EXPINF10YR",           # Cleveland Fed: 10-year expected inflation, monthly
    "REAINTRATREARAT10Y",   # Cleveland Fed: 10-year real rate (incl. real risk premium), monthly
    "TENEXPCHAREARISPRE",   # Cleveland Fed: 10-year real risk premium, monthly
    "TENEXPCHAINFRISPRE",   # Cleveland Fed: 10-year inflation risk premium, monthly
]

raw_dir = ROOT / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)

for sid in series_ids:
    s = fred.get_series(sid)
    s.name = sid
    s.index.name = "date"
    s.to_csv(raw_dir / f"{sid}.csv")
    print(f"{sid}: {s.index.min().date()} to {s.index.max().date()}, {len(s)} rows, {s.isna().sum()} missing")
```
