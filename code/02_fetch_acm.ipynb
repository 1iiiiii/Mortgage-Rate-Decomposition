# 02 Fetch the ACM term premium from the NY Fed

Download the NY Fed ACM term-premium file and save the 10-year columns to `data/raw/acm.csv`.


```python
from pathlib import Path
import io
import requests
import pandas as pd

ROOT = Path.cwd().parent        # repository root; all paths build from here
```

```python
URL = "https://www.newyorkfed.org/medialibrary/media/research/data_indicators/ACMTermPremium.xls"

# Scrape setup
r = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"}, timeout=60)
r.raise_for_status()

# Get daily frequency
acm = pd.read_excel(io.BytesIO(r.content), sheet_name="ACM Daily")

# ACMY10 = fitted 10y yield, ACMRNY10 = risk-neutral yield (expected short rate), ACMTP10 = term premium
acm = acm[["DATE", "ACMY10", "ACMRNY10", "ACMTP10"]]
acm["DATE"] = pd.to_datetime(acm["DATE"], format="%d-%b-%Y")
acm = acm.rename(columns={"DATE": "date"}).set_index("date")

raw_dir = ROOT / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)
acm.to_csv(raw_dir / "acm.csv")

print(f"{acm.index.min().date()} to {acm.index.max().date()}, {len(acm)} rows")
```
